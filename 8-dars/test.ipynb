#  1-AMALIYOT: TOKENIZATION & SCALING LAWS (GPT-2 VS QWEN-2.5)
**Maqsadi:** Klassik GPT-2 va Zamonaviy Qwen-2.5 (151k Vocab) tokenizerlarini turli tillarda (Ingliz, O'zbek), Kodda va Emojilarda taqqoslash.


```python
# Zaruriy kutubxonalarni o'rnatish va import qilish
!pip install -q transformers sentencepiece torch accelerate

import torch
from transformers import AutoTokenizer

print(f"PyTorch versiyasi: {torch.__version__}")
print(f"CUDA (GPU) mavjudmi?: {torch.cuda.is_available()}")
```

### 2-BOSQICH: Ikki Xil Avlod Tokenizerlarini Yuklash
- **GPT-2 Tokenizer:** 50,257 ta lug'atga ega klassik Byte-level BPE.
- **Qwen2.5 Tokenizer:** 151,646 ta lug'atga ega zamonaviy ko'p tilli BPE.


```python
# Tokenizerlarni yuklash
model_gpt2 = "gpt2"
model_qwen = "Qwen/Qwen2.5-0.5B"

tokenizer_gpt2 = AutoTokenizer.from_pretrained(model_gpt2)
tokenizer_qwen = AutoTokenizer.from_pretrained(model_qwen)

print(f"GPT-2 Lug'at hajmi (Vocab Size): {tokenizer_gpt2.vocab_size:,}")
print(f"Qwen2.5 Lug'at hajmi (Vocab Size): {tokenizer_qwen.vocab_size:,}")
```

### 3-BOSQICH: O'zbek Tili Matnlarida Subword Bo'linishini Taqqoslash
O'zbekcha matnlar turli tokenizerlarda qanchalik maydalanib ketishini ko'ramiz.


```python
# O'zbekcha matn amaliyoti
text_uz = "Sun'iy intellekt va neyron tarmoqlari kelajak texnologiyasidir."

tokens_gpt2 = tokenizer_gpt2.tokenize(text_uz)
tokens_qwen = tokenizer_qwen.tokenize(text_uz)

print(f"--- MATN: '{text_uz}' ---\n")
print(f"GPT-2 Tokenlar soni ({len(tokens_gpt2)} ta): {tokens_gpt2}")
print(f"Qwen2.5 Tokenlar soni ({len(tokens_qwen)} ta): {tokens_qwen}")
```

### 4-BOSQICH: Kod (Python) va Matematik Matnlarda Tokenizatsiya
AI modellari dasturlash kodlarini qanday subwordlarga bo'lishini tekshiramiz.


```python
#  Python kodi va Matematika
code_sample = "def calculate_loss(y_true, y_pred):\n    return np.mean((y_true - y_pred) ** 2)"

print("--- PYTHON KODI TOKENIZATSIYASI ---")
print("GPT-2 Tokenlar:", tokenizer_gpt2.tokenize(code_sample))
print("Qwen2.5 Tokenlar:", tokenizer_qwen.tokenize(code_sample))

print("\n--- TOKEN ID VA DECODE TESTI (Qwen2.5) ---")
ids = tokenizer_qwen.encode(code_sample)
decoded = tokenizer_qwen.decode(ids)
print("Token ID lar:", ids[:10], "...")
print("Qayta tiklangan matn (Decoded):", decoded)
```

### 5-BOSQICH: Emojilar va Maxsus Simvollar Testi
Byte-level BPE emojilarni qanday baytlarga bo'lishini kuzatamiz.


```python
#  Emoji va Multi-byte character test
emoji_text = "Neyron tarmoq 🚀 GPT-4 🤖 2026-yil"

print("GPT-2 Emoji Tokenlar:", tokenizer_gpt2.tokenize(emoji_text))
print("Qwen2.5 Emoji Tokenlar:", tokenizer_qwen.tokenize(emoji_text))
```

### 6-BOSQICH: Chinchilla Scaling Law Hisob-kitobi (Taqqoslash)
Chinchilla qoidasiga ko'ra ($D \approx 20N$):
1. **GPT-2 Small (124M parametr)**
2. **Qwen2.5-0.5B (490M parametr)**
Ushbu modellarni optimal o'qitish uchun qancha token va necha GB dataset kerakligini hisoblaymiz.


```python
# Cell 6: Chinchilla Scaling Law Kalkulyatori

def calculate_chinchilla(param_count_millions, model_name):
    params = param_count_millions * 1e6
    tokens_required = params * 20  # D = 20 * N
    bytes_required = tokens_required * 4  # 1 token ≈ 4 bytes
    gb_required = bytes_required / (1024 ** 3)
    
    print(f"=== {model_name} ({param_count_millions}M Parametr) ===")
    print(f" Talab qilinadigan Optimal Tokenlar: {tokens_required / 1e9:.2f} Billion Tokens")
    print(f" Zaruriy Matnli Dataset Hajmi: {gb_required:.2f} GB\n")

calculate_chinchilla(124, "GPT-2 Small")
calculate_chinchilla(490, "Qwen2.5-0.5B")
calculate_chinchilla(1500, "Llama-3.2-1B / Gemma-2-2B")
```
